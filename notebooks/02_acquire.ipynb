# 2. Acquire and Process Data

This notebook demonstrates how to populate the LC-MS pipeline with simulated data.

```python
from lcms_demo.pipeline import subject, session, scan
```

## Generic Demo Data

Generate a simple demo dataset with configurable parameters.

```python
from lcms_demo.simulation import acquire_demo_data

summary = acquire_demo_data(
    n_subjects=3,
    samples_per_subject=2,
    scans_per_session=50,
    seed=42,
)

print(f"Generic demo data acquired:")
print(f"  - {summary['subjects']} subjects")
print(f"  - {summary['samples']} samples")
print(f"  - {summary['sessions']} sessions")
```

## NVS-4821 Hepatotoxicity Study

A realistic preclinical study with treatment groups and time-course sampling.

```python
from lcms_demo.simulation import acquire_nvs4821_study

summary = acquire_nvs4821_study(
    n_scans_per_session=50,
    seed=42,
)

print(f"NVS-4821 Study acquired:")
print(f"  - {summary['subjects']} subjects")
print(f"  - {summary['samples']} samples")
print(f"  - {summary['sessions']} sessions")
```

## Populate Pipeline

Run the pipeline's `make()` methods to process all sessions.

```python
scan.Acquisition.populate(display_progress=True)
scan.MassAnalysis.populate(display_progress=True)
scan.PeakDetection.populate(display_progress=True)

print("\nPipeline populated!")
```
