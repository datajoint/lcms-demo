# 3. Populate the Pipeline

This notebook demonstrates how to populate the LC-MS pipeline with simulated data.

```python
from lcms_demo.pipeline import subject, session, scan
```

## Option 1: Generic Demo Data

Generate a simple demo dataset with configurable parameters.

```python
from lcms_demo.simulation import acquire_demo_data

# Step 1: Acquire Manual tables (Subject, Sample, Session)
summary = acquire_demo_data(
    n_subjects=3,
    samples_per_subject=2,
    scans_per_session=50,
    seed=42,
)

print(f"Manual tables acquired:")
print(f"  - {summary['subjects']} subjects")
print(f"  - {summary['samples']} samples")
print(f"  - {summary['sessions']} sessions")
```

```python
# Step 2: Populate Imported/Computed tables by calling .populate()
# This triggers the make() methods which read from the data source

scan.Scans.populate(display_progress=True)
scan.Spectra.populate(display_progress=True)
scan.Peaks.populate(display_progress=True)

print("\nPipeline populated!")
```

## Manual Data Entry

You can also insert data manually using DataJoint's insert methods.

```python
# Insert a single subject
subject.Subject.insert1(
    {"subject_id": "MANUAL_001", "subject_description": "Manually entered subject"},
    skip_duplicates=True,
)

subject.Subject()
```

```python
# Insert multiple samples
subject.Sample.insert(
    [
        {"subject_id": "MANUAL_001", "sample_id": "plasma_01", "sample_type": "plasma"},
        {"subject_id": "MANUAL_001", "sample_id": "liver_01", "sample_type": "liver"},
    ],
    skip_duplicates=True,
)

subject.Sample()
```
